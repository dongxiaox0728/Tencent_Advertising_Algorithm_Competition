# UNI-REC Block 设计说明

## 背景

本赛题要求设计一个统一backbone，同时处理两类特征：
- **S（Sequential）**：用户历史行为序列，具有时序性
- **NS（Non-Sequential）**：用户/物品静态属性，异构多字段

现有方案（如OneTrans）直接把S和NS拼在一起做统一self-attention，问题在于S token和NS token语义上完全不同，强行对等交互会引入噪声。

---

## 核心设计思想

S和NS之间应该是**相互指导、动态制约**的关系：
- S反映用户近期兴趣，应该能修正NS的静态先验
- NS代表用户长期属性，应该能约束S的短期波动
- 两个方向的影响强度**不应固定**，而应根据当前样本内容自适应决定

---

## Block结构（三步顺序执行）

```
输入: s [B, L_s, D],  ns [B, L_ns, D]
        ↓
  Step 1: S内部 — Masked Self-Attention
        （捕捉序列内时序依赖，支持causal/bidirectional两种模式）
        ↓
  Step 2: NS内部 — Per-Token FFN
        （每个字段独立变换，不做跨字段attention，
          因为NS各字段语义异构，跨字段attention无意义）
        ↓
  Step 3: S × NS — 自适应双向Cross-Attention
        （见下方详细说明）
        ↓
输出: s [B, L_s, D],  ns [B, L_ns, D]
```

先做内部处理，是为了让S和NS各自形成稳定的表示，再用这个"成熟"的表示去做跨类型交互，gate信号的质量更高。

---

## 自适应双向Cross-Attention

这是本方案的核心创新点。

**两个方向同时计算：**
- S → NS：S作Q，NS作KV（S从静态属性中提取相关信息）
- NS → S：NS作Q，S作KV（NS从近期行为中更新自身表示）

**向量Gate动态控制方向：**

$$g = \sigma\left(W \cdot [h_S;\ h_{NS};\ h_S \odot h_{NS}]\right) \in \mathbb{R}^D$$

其中 $h_S, h_{NS}$ 是S和NS的mean pooling，$h_S \odot h_{NS}$ 是逐元素乘积（捕捉每个语义维度上两侧的一致性或分歧）。

**最终更新：**

$$S' = \text{Norm}(S + g \odot \text{CrossAttn}(S \to NS))$$
$$NS' = \text{Norm}(NS + (1-g) \odot \text{CrossAttn}(NS \to S))$$

$g$ 接近1时，S大量吸收NS信息，NS侧更新被压制；$g$ 接近0时反过来。gate由当前样本的内容决定，不是固定参数。

---

## 整体Backbone

```
Embedding层（待对接baseline数据pipeline）
        ↓
UniRECBlock × N层（可堆叠）
        ↓
Mean Pool S + Mean Pool NS → 拼接
        ↓
MLP → 输出logit（BCE loss训练）
```

N层数量、hidden dim、causal mask开关均通过config控制，方便做scaling实验。

---

## 与OneTrans的关键区别

| | OneTrans | 本方案 |
|---|---|---|
| S/NS交互方式 | 统一self-attention，所有token地位平等 | 双向cross-attention，类型感知 |
| 交互方向 | 对称 | 自适应，由内容动态决定 |
| NS内部处理 | self-attention | per-token FFN |
| S内部处理 | self-attention | masked self-attention |

---

## 代码文件

完整PyTorch实现，包含`UniRECBlock`和`UniRECBackbone`，smoke test已通过。

待补全：embedding层、padding mask、三条序列（action_seq/content_seq/item_seq）的合并方式——等Round 1 baseline代码释放后对接。

````python
import torch
import torch.nn as nn
import torch.nn.functional as F
import math


class MultiHeadAttention(nn.Module):
    def __init__(self, d_model: int, n_heads: int, dropout: float = 0.1):
        super().__init__()
        assert d_model % n_heads == 0
        self.d_model = d_model
        self.n_heads = n_heads
        self.d_head = d_model // n_heads

        self.q_proj = nn.Linear(d_model, d_model)
        self.k_proj = nn.Linear(d_model, d_model)
        self.v_proj = nn.Linear(d_model, d_model)
        self.out_proj = nn.Linear(d_model, d_model)
        self.dropout = nn.Dropout(dropout)

    def _split_heads(self, x: torch.Tensor) -> torch.Tensor:
        # [B, L, D] -> [B, H, L, d_head]
        B, L, D = x.shape
        return x.view(B, L, self.n_heads, self.d_head).transpose(1, 2)

    def _merge_heads(self, x: torch.Tensor) -> torch.Tensor:
        # [B, H, L, d_head] -> [B, L, D]
        B, H, L, d = x.shape
        return x.transpose(1, 2).contiguous().view(B, L, H * d)

    def forward(
        self,
        q: torch.Tensor,
        k: torch.Tensor,
        v: torch.Tensor,
        mask: torch.Tensor = None,
    ) -> torch.Tensor:
        Q = self._split_heads(self.q_proj(q))
        K = self._split_heads(self.k_proj(k))
        V = self._split_heads(self.v_proj(v))

        scale = math.sqrt(self.d_head)
        attn = torch.matmul(Q, K.transpose(-2, -1)) / scale  # [B, H, Lq, Lk]

        if mask is not None:
            attn = attn.masked_fill(mask, float('-inf'))

        attn = self.dropout(F.softmax(attn, dim=-1))
        out = torch.matmul(attn, V)  # [B, H, Lq, d_head]
        return self.out_proj(self._merge_heads(out))


class PerTokenFFN(nn.Module):
    """
    Per-token FFN for NS tokens: each token is processed independently,
    no cross-token interaction (consistent with RankMixer's design rationale
    — NS tokens are heterogeneous fields, cross-field attention is meaningless).
    """
    def __init__(self, d_model: int, d_ff: int, dropout: float = 0.1):
        super().__init__()
        self.ffn = nn.Sequential(
            nn.Linear(d_model, d_ff),
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(d_ff, d_model),
            nn.Dropout(dropout),
        )
        self.norm = nn.LayerNorm(d_model)

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        # x: [B, L_ns, D] — applied token-wise, no interaction across tokens
        return self.norm(x + self.ffn(x))


class AdaptiveBidirectionalCrossAttention(nn.Module):
    """
    Adaptive bidirectional cross-attention between S and NS tokens.

    Both directions are computed:
      - S attends to NS  (S as Q, NS as K/V)
      - NS attends to S  (NS as Q, S as K/V)

    A vector gate g in R^D is computed from the aggregated representations
    of both sides via cosine similarity + a learned linear transform.
    The gate controls how much each direction contributes to the final update.

    gate g close to 1 -> S-guided update dominates (recent behavior drives NS)
    gate g close to 0 -> NS-guided update dominates (static profile drives S)
    """
    def __init__(self, d_model: int, n_heads: int, dropout: float = 0.1):
        super().__init__()
        # S -> NS direction: S as Q, NS as K/V
        self.s_to_ns_attn = MultiHeadAttention(d_model, n_heads, dropout)
        # NS -> S direction: NS as Q, S as K/V
        self.ns_to_s_attn = MultiHeadAttention(d_model, n_heads, dropout)

        # Vector gate: input is [h_s_pool; h_ns_pool; cosine_sim_expanded]
        # We concatenate both pooled representations + element-wise product
        # as a proxy for fine-grained agreement per dimension
        self.gate_proj = nn.Sequential(
            nn.Linear(d_model * 3, d_model),
            nn.Sigmoid(),
        )

        self.norm_s = nn.LayerNorm(d_model)
        self.norm_ns = nn.LayerNorm(d_model)
        self.dropout = nn.Dropout(dropout)

    def _pool(self, x: torch.Tensor) -> torch.Tensor:
        # Mean pool over sequence dimension: [B, L, D] -> [B, D]
        return x.mean(dim=1)

    def forward(
        self,
        s: torch.Tensor,    # [B, L_s, D]
        ns: torch.Tensor,   # [B, L_ns, D]
    ):
        h_s = self._pool(s)    # [B, D]
        h_ns = self._pool(ns)  # [B, D]

        # Element-wise product captures per-dimension agreement/disagreement
        elem_prod = h_s * h_ns  # [B, D]

        gate_input = torch.cat([h_s, h_ns, elem_prod], dim=-1)  # [B, 3D]
        g = self.gate_proj(gate_input).unsqueeze(1)  # [B, 1, D]

        # S attends NS: captures how static profile shapes sequence understanding
        s_cross = self.s_to_ns_attn(q=s, k=ns, v=ns)   # [B, L_s, D]
        # NS attends S: captures how recent behavior updates static profile
        ns_cross = self.ns_to_s_attn(q=ns, k=s, v=s)   # [B, L_ns, D]

        # g controls direction: g~1 means S-side update dominates (S absorbs NS info)
        # g~0 means NS-side update dominates (NS absorbs S info)
        s_out = self.norm_s(s + self.dropout(g * s_cross))
        ns_out = self.norm_ns(ns + self.dropout((1 - g) * ns_cross))

        return s_out, ns_out


class UniRECBlock(nn.Module):
    """
    A single stackable UNI-REC block.

    Processing order within one block:
      1. S internal: masked self-attention (causal or bidirectional, config-controlled)
      2. NS internal: per-token FFN (no cross-field attention)
      3. S x NS cross: adaptive bidirectional cross-attention with vector gate

    This order ensures both S and NS have stable representations
    before cross-type interaction — the gate signal is computed from
    'mature' representations rather than raw inputs.
    """
    def __init__(
        self,
        d_model: int,
        n_heads: int,
        d_ff: int,
        dropout: float = 0.1,
        causal_mask: bool = True,   # True: autoregressive, False: bidirectional
    ):
        super().__init__()
        self.causal_mask = causal_mask

        # Step 1: S internal self-attention
        self.s_self_attn = MultiHeadAttention(d_model, n_heads, dropout)
        self.s_self_norm = nn.LayerNorm(d_model)
        self.s_ffn = nn.Sequential(
            nn.Linear(d_model, d_ff),
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(d_ff, d_model),
            nn.Dropout(dropout),
        )
        self.s_ffn_norm = nn.LayerNorm(d_model)

        # Step 2: NS internal per-token FFN
        self.ns_internal = PerTokenFFN(d_model, d_ff, dropout)

        # Step 3: adaptive bidirectional cross-attention
        self.cross_attn = AdaptiveBidirectionalCrossAttention(d_model, n_heads, dropout)

    def _causal_mask(self, seq_len: int, device: torch.device) -> torch.Tensor:
        # Upper triangular mask (excluding diagonal): True means masked out
        return torch.triu(
            torch.ones(seq_len, seq_len, dtype=torch.bool, device=device), diagonal=1
        )

    def forward(
        self,
        s: torch.Tensor,    # [B, L_s, D]
        ns: torch.Tensor,   # [B, L_ns, D]
    ):
        # --- Step 1: S internal ---
        mask = None
        if self.causal_mask:
            mask = self._causal_mask(s.size(1), s.device)  # [L_s, L_s]
            # broadcast to [1, 1, L_s, L_s] for multi-head attention
            mask = mask.unsqueeze(0).unsqueeze(0)

        s_attn_out = self.s_self_attn(q=s, k=s, v=s, mask=mask)
        s = self.s_self_norm(s + s_attn_out)
        s = self.s_ffn_norm(s + self.s_ffn(s))

        # --- Step 2: NS internal ---
        ns = self.ns_internal(ns)

        # --- Step 3: cross-type interaction ---
        s, ns = self.cross_attn(s, ns)

        return s, ns


class UniRECBackbone(nn.Module):
    """
    Full backbone: stack N UniRECBlocks, then pool and predict.
    All blocks share the same causal_mask setting (config-controlled).
    """
    def __init__(
        self,
        d_model: int = 128,
        n_heads: int = 4,
        d_ff: int = 256,
        n_layers: int = 3,
        dropout: float = 0.1,
        causal_mask: bool = True,
    ):
        super().__init__()
        self.blocks = nn.ModuleList([
            UniRECBlock(d_model, n_heads, d_ff, dropout, causal_mask)
            for _ in range(n_layers)
        ])
        # Final prediction head: pool S and NS, concat, project to logit
        self.head = nn.Sequential(
            nn.Linear(d_model * 2, d_model),
            nn.GELU(),
            nn.Linear(d_model, 1),
        )

    def forward(
        self,
        s: torch.Tensor,    # [B, L_s, D]
        ns: torch.Tensor,   # [B, L_ns, D]
    ) -> torch.Tensor:
        for block in self.blocks:
            s, ns = block(s, ns)

        # Pool both sides and concat for final prediction
        s_pool = s.mean(dim=1)    # [B, D]
        ns_pool = ns.mean(dim=1)  # [B, D]
        out = self.head(torch.cat([s_pool, ns_pool], dim=-1))  # [B, 1]
        return out.squeeze(-1)   # [B]


if __name__ == '__main__':
    torch.manual_seed(42)
    B, L_s, L_ns, D = 4, 20, 15, 128

    s = torch.randn(B, L_s, D)
    ns = torch.randn(B, L_ns, D)

    print("=== UniRECBlock (causal) ===")
    block_causal = UniRECBlock(d_model=D, n_heads=4, d_ff=256, causal_mask=True)
    s_out, ns_out = block_causal(s, ns)
    print(f"s_out: {s_out.shape}, ns_out: {ns_out.shape}")

    print("\n=== UniRECBlock (bidirectional) ===")
    block_bi = UniRECBlock(d_model=D, n_heads=4, d_ff=256, causal_mask=False)
    s_out, ns_out = block_bi(s, ns)
    print(f"s_out: {s_out.shape}, ns_out: {ns_out.shape}")

    print("\n=== UniRECBackbone (3 layers, causal) ===")
    model = UniRECBackbone(d_model=D, n_heads=4, d_ff=256, n_layers=3, causal_mask=True)
    logits = model(s, ns)
    print(f"logits: {logits.shape}")

    total = sum(p.numel() for p in model.parameters())
    print(f"total params: {total:,}")

    print("\n=== gradient check ===")
    loss = logits.sum()
    loss.backward()
    print("backward pass: OK")
````

````python
print("=== 序列时间跨度分析 ===")
# content_seq的feature_id=41看起来是时间戳序列
def get_ts_seq(row):
    for f in row['seq_feature']['content_seq']:
        if f['feature_id'] == 41:
            return f['int_array']
    return None

spans_days = []
gaps_hours = []
for _, row in df.iterrows():
    ts = get_ts_seq(row)
    if ts is not None and len(ts) > 1:
        ts_sorted = sorted(ts, reverse=True)
        span = (ts_sorted[0] - ts_sorted[-1]) / 86400
        spans_days.append(span)
        diffs = [abs(ts_sorted[i]-ts_sorted[i+1])/3600
                 for i in range(min(10, len(ts_sorted)-1))]
        gaps_hours.extend(diffs)

print(f"content_seq 时间跨度:")
print(f"  min={min(spans_days):.1f}天, max={max(spans_days):.1f}天, "
      f"mean={np.mean(spans_days):.1f}天")
print(f"相邻行为间隔 (前10步, 小时):")
print(f"  min={min(gaps_hours):.2f}h, max={max(gaps_hours):.1f}h, "
      f"mean={np.mean(gaps_hours):.2f}h, median={np.median(gaps_hours):.2f}h")

# 正负样本的序列长度是否有差异
pos_idx = df['label'].apply(lambda x: x[0]['action_type']) == 2
neg_idx = ~pos_idx
print("=== 正负样本序列长度对比 ===")
pos_idx = df['label'].apply(lambda x: x[0]['action_type']) == 2
for seq_key in ['action_seq', 'content_seq', 'item_seq']:
    pos_lens, neg_lens = [], []
    for i, (_, row) in enumerate(df.iterrows()):
        seq = row['seq_feature'][seq_key]
        if len(seq) == 0 or seq[0]['int_array'] is None:
            continue
        l = len(seq[0]['int_array'])
        if pos_idx.iloc[i]:
            pos_lens.append(l)
        else:
            neg_lens.append(l)
    print(f"{seq_key}: 正样本均值={np.mean(pos_lens):.1f}, "
          f"负样本均值={np.mean(neg_lens):.1f}")
````

````python
print("=== item_seq 与 target item_id 重叠 ===")
overlap_count = 0
valid_count = 0
for _, row in df.iterrows():
    seq = row['seq_feature']['item_seq']
    if len(seq) == 0 or seq[0]['int_array'] is None:
        continue
    hist_items = set(seq[0]['int_array'].tolist())
    target = row['item_id']
    valid_count += 1
    if target in hist_items:
        overlap_count += 1

print(f"有效样本: {valid_count}, target出现在item_seq中: "
      f"{overlap_count} ({overlap_count/valid_count*100:.1f}%)")

print("\n=== sample timestamp vs 序列最近行为时间 ===")
diffs = []
for _, row in df.iterrows():
    seq = row['seq_feature']['content_seq']
    if len(seq) == 0 or seq[0]['int_array'] is None:
        continue
    for f in seq:
        if f['feature_id'] == 41:
            ts = f['int_array']
            most_recent = int(max(ts))
            diff = (row['timestamp'] - most_recent) / 3600
            diffs.append(diff)
            break

print(f"sample timestamp - 序列最近行为时间:")
print(f"  mean={np.mean(diffs):.2f}h, median={np.median(diffs):.2f}h")
print(f"  min={min(diffs):.2f}h, max={max(diffs):.2f}h")
print(f"  负值比例: {sum(d<0 for d in diffs)/len(diffs)*100:.1f}%")
````
